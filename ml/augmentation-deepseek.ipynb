```python
from langchain_deepseek import ChatDeepSeek


llm = ChatDeepSeek(
    model="deepseek-chat",
    temperature=0.5,
    max_tokens=None,
    timeout=None,
    max_retries=2,
    # other params...
)

gen_prompt = lambda sentence: f'''Ты чат-бот поддержки рутуб, на какие 5 вопросов пользователя ты бы ответил с помощью {sentence}. Верни ТОЛЬКО 5 вопросов в формате JSON'''
```

```python
import time
augmented_questions = []
for id, row in faq_csv.iterrows():
    if id <= 263:
        continue
    begin = time.time()
    question = row["question"]
    questions = llm.invoke(gen_prompt(question)).text()
    augmented_questions.append({"id": id, "questions": questions})
    end = time.time()
    print(f"Finished iteration for id={id} in {end - begin} seconds") 
```

```python
import pickle

with open("augmented_questions.pk", "wb") as f:
    pickle.dump(augmented_questions, f)
len(augmented_questions)
```
