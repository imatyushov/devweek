````python
faq_path = "databases/01_FAQ.csv"
stubs_path = "databases/02_Stubs.csv"

import pandas as pd

faq_csv = pd.read_csv(faq_path, delimiter="|")
faq_csv.head()
````

````python
import pickle
with open('augmented_questions.pk', 'rb') as f:
    questions=pickle.load(f)
questions[0]
````

````python
import json

dataset = []

misses = 0
for question in questions:
    try:
        id = question["id"]
        json_questions = json.loads(question["questions"].replace("```json", "").replace("```", ""))
        questions_list = [item["question"] for item in json_questions]
        row = faq_csv.iloc[id]
        dataset.append((row["category"], row["question"], row["answer"]))
        for _question in questions_list:
            result = (row["category"], _question, row["answer"])
            dataset.append(result)
    except:
        misses += 1
print(misses)
print(dataset[0])
````

````python
import csv


with open("augmented_FAQ.csv", 'w') as out:
    csv_out = csv.writer(out, delimiter="|")
    csv_out.writerow(["category", "question", "answer"])
    csv_out.writerows(dataset)
````
